```python
import numpy as np
from numpy import linalg as LA
import pandas as pd

def companion_matrix(coeffs):
    n = len(coeffs) - 1
    C = np.zeros((n,n))
    C[range(1,n), range(n-1)] = 1 #subdiagonal of 1s
    C[:, -1] = -np.array(coeffs[:0:-1]) # last col = -[a_{n-1},...,a_0]     -np is required for proper ordering
    return C
```

1. Find the companion matrices for the following polynomials.

a) p(t)=t^2+2t-4

b) p(t)=t^3-9t^2 +12t+22

c) p(t)=t^4-2t^3 - 13t^2 +14t +24

d) Find the characteristic polynomials of the matrices you just found in parts a-c.

```python
p_a = [1, 2, -4]
C_a = companion_matrix(p_a)
print("Companion matrix for p(t) = t^2 +2t - 4:\n", C_a)

print()

p_b = [1, -9, 12, 22]
C_b = companion_matrix(p_b)
print("Companion matrix for p(t) = t^3 -9t^2 +12t + 22:\n", C_b)

print()

p_c = [1, -2, -13, 14, 24]
C_c = companion_matrix(p_c)
print("Companion matrix for p(t) = t^4 -2t^3 -13t^2 + 14t + 24:\n", C_c)

char_a = np.poly(C_a)
char_b = np.poly(C_b)
char_c = np.poly(C_c)

print()
print("Characteristic polynomial of C_a:", char_a)
print("Characteristic polynomial of C_b:", char_b)
print("Characteristic polynomial of C_c:", char_c)
```

2. Show that the characteristic polynomial of a companion matrix for the n
th degree polynomial p(t) is det(C - I_n) = (-1)^n p(λ)  as follows. 
a) Show that if C is the companion matrix for a quadratic polynomial p(t)= t^2 +at+a_0, then det(C - I_n) = (-1)^n p(λ) by direct computation.

b) Use mathematical induction to show that the result holds for n ≥ 2 . Hint: expand the necessary determinant by cofactors down the first column.

a. 
Let p(t) = t^2 +at+a_0

Companion Matrix = C = [[0 -a0
                         1 -a1]]

Then:              C -λI_2 = [[-λ  -a0
                                1  -a1 - λ]]

Determinant:

                det(C -λI_2) = (-λ)(-a1 - λ) - (-a0)(1)
                             = λ(λ + a1) + a0
                             = λ^2 + a1λ +a0 = p(λ)

for n = 2: det(C -λI_n) = p(λ) = (-1)^2 p(λ)

b. 
Base Case: n = 2
proven in a.

Assume the result holds for degree n -1: det(C_n-1 -λI_n-1) = (-1)^n-1 p_n-1(λ)

where C_n-1 is the (n - 1) x (n - 1) companion matrix

Inductive Step:
For degree n, the companion matrix is:

C =  [[ 0.   0 ... 0 -a0]
     [  1.   0 ... 0 -a1]
     [  0.   1 ... 0 -a2]
     [  0.   0 ... 1 -a_n-1]]

Then C-λI = [[ -λ.   0 ... 0 -a0]
            [   1.  -λ ... 0 -a1]
            [   0.   1 ... 0 -a2]
            [   0.   0 ... 1 -a_n-λ]]

Step 1: Expand along the first Column
        det(C -λI_n) = (-λ)det(M_1) + (-1)^(2+1) (1) det(M_2)

Where:
- M_1 is the (n - 1) x (n - 1) matrix after removing first row and first column.
- M_2 is the (n - 1) x (n - 1) matrix after removing secound row and first column.

Cofactor expandsion:

You get a recursion: det(C_n -λI_n) = (-λ)det(C_n-1 -λI_n-1) - a0

Expanding recursively reproduces the polynomial: 

det(C_n -λI_n) = (-1)^n (λ^n + a_n-1 λ^n-1+...+a0) = (-1)^n p(λ)

4. Approximate the roots of the polynomials in Question 1 by applying the QR Method to their companion matrices. Iterate until the entries below the main diagonal are all below 0.1

QR Methods to companion matrices: C_a, C_b, C_c

```python
def qr_eigenvalues(A, tol=0.1,max_iter=1000):
    
    n = A.shape[0]
    Ak = A.copy()
    for i in range(max_iter):
        Q, R = np.linalg.qr(Ak)
        Ak = R @ Q
        # Check subdiagonal entries
        if np.all(np.abs(np.tril(Ak, -1)) < tol):
            break
    return np.diag(Ak), Ak # approximate eigenvalues, final matrix

#Apply to all companion matrices
eig_a, final_a = qr_eigenvalues(C_a)
eig_b, final_b = qr_eigenvalues(C_b)
eig_c, final_c = qr_eigenvalues(C_c)

print("Approximate roots of p_a:", eig_a)
print("Approximate roots of p_b:", eig_b)
print("Approximate roots of p_c:", eig_c)
```
