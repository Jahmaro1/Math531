```python
import numpy as np
from numpy import linalg as LA
import pandas as pd
```

Create the Matrix A

```python
I = np.eye(3) # 3x3 identity matrix
A = np.array([
    [6,-4,-8],
    [-3,10,11],
    [2,-4,-4]
    ])
print(A)
```

Compute A^10

```python
# A_Raised = A^10
A_10 = np.linalg.matrix_power(A,10)
print(A_10)
```

Compute (A/6) ^ 10

```python
A_Div6 = np.linalg.matrix_power((A/6),10)
print(A_Div6)
```

Explore what happens when A is powered up even higher by increasing the exponent. What do
you notice when you compare different high powers of A/6?

```python
# for i in range(10):
#     print("A^" + str(i+10))
#     print( (A//6)^(i+10))

for i in range(10,110,10):
    print("A^" + str(i))
    print(np.linalg.matrix_power((A/6),i))
```
